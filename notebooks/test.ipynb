```python
import requests
import json
import os
import sys
sys.path.append("../src")

from config import FREECRYPTO_TOKEN,NEWS_API_KEY
```

```python
#Autorization header
headers = {
    "Authorization": f"Bearer {FREE_CRYPTO_TOKEN}"
}
```

```python
#endpoint URL
url = "https://api.freecryptoapi.com/v1/getCryptoList"
```

```python
#send request
response = requests.get(url,headers=headers)
data = response.json()
print(json.dumps(data,indent=4))
```

```python
url = "https://api.freecryptoapi.com/v1/getData"
params = {"symbol":"BTC"}
```

```python
#request with params
response = requests.get(url,headers=headers,params = params)
data = response.json()
print(json.dumps(data,indent=4))
```
